```python
#Tratamiento de datos
import pandas as pd

pd.set_option('display.max_columns',None)

#Exploracion de archivos
import sys
sys.path.append('../')

#funciones personalizadas
from SRC import sp_eda as se
from SRC import sp_limpieza as sl
```

```python
df_u = pd.read_csv('../Datos/df_data_unificada.csv')
```

AJUSTE NOMBRE COLUMNAS

```python

#Cambion de nombre de columnas (espacios y puntos por "_")
df_u.columns = df_u.columns.str.replace(' ', '_')
df_u.columns = df_u.columns.str.replace('.', '_')
df_u.columns = df_u.columns.str.lower()
```

```python
df_u.columns
```

### LIMPIEZA

ELIMINACION DE COLUMNAS

```python
df_u = df_u.drop(columns=['unnamed:_0_x', 'unnamed:_0_y', 'id_'])
```

AGE

```python
#Conversion de float a Integer
df_u['age'] = df_u['age'].astype('Int64')
```

N.EMPLEADOS

```python
#cambio "," por "." y conversion de Object a integer
df_u['nr_employed'] = df_u['nr_employed'].str.replace(',', '.').str.strip()
df_u['nr_employed'] = df_u['nr_employed'].astype(float).astype(int)
```

```python
df_u['job'].unique()
```

JOB

```python
# Cambio "-" por "_" y eliminacion de "." al final de admin
df_u['job'] = df_u['job'].str.replace('-', '_').str.strip('.')  # quitar espacios
```

FECHAS

```python
df_u.sample(5)
```

```python
meses = {
    'enero':'01', 'febrero':'02', 'marzo':'03', 'abril':'04',
    'mayo':'05', 'junio':'06', 'julio':'07', 'agosto':'08',
    'septiembre':'09', 'octubre':'10', 'noviembre':'11', 'diciembre':'12'
}

for mes_texto, mes_num in meses.items():
    df_u['date'] = df_u['date'].str.replace(mes_texto, mes_num)


for col in ['date', 'dt_customer']:
    df_u[col] = pd.to_datetime(df_u[col], dayfirst=True, errors='coerce')
```

```python
cols = ['default', 'housing', 'loan']

for col in cols:

   df_u[col + '_str'] = df_u[col].astype(object).replace({0.0: 'No', 1.0: 'Yes'})
```
