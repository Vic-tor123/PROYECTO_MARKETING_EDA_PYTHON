comprobamos si todos los elementos de df_e estan en df_c 

```python
ids_c = set(df_c['id_'])
ids_e = set(df_e['ID'])
```

```python
ids_no_en_e = ids_c - ids_e
ids_no_en_e_series = pd.Series(list(ids_no_en_e))
Num_ids_no_en_e = len(ids_no_en_e_series)
Num_ids_no_en_e
```

```python
df_2012 = pd.read_excel('../Datos.Originales/customer-details.xlsx', sheet_name ='2012')
df_2013 = pd.read_excel('../Datos.Originales/customer-details.xlsx', sheet_name ='2013')
df_2014 = pd.read_excel('../Datos.Originales/customer-details.xlsx', sheet_name ='2014')
```

```python
df_años_unificado = pd.concat([df_2012,df_2013,df_2014], ignore_index=True)
```

```python
ids_c = set(df_c['id_'])
ids_u = set(df_años_unificado['ID'])
```

```python
ids_no_en_u = ids_c - ids_u
ids_no_en_u_series = pd.Series(list(ids_no_en_u))
Num_ids_no_en_u = len(ids_no_en_u_series)
Num_ids_no_en_u
```

```python
ids_no_en_c = ids_u - ids_c
ids_no_en_c_series = pd.Series(list(ids_no_en_c))
Num_ids_no_en_c = len(ids_no_en_c_series)
Num_ids_no_en_c
```

```python

df_años_unificado.sample(5)
```

```python
# Contar duplicados en df_c['id_']
num_duplicados_df_c = df_c['id_'].duplicated(keep=False).sum()
print(f"Número de duplicados en df_c['id']: {num_duplicados_df_c}")

# Contar duplicados en df_años_unificado['ID']
num_duplicados_df_años_unificado = df_años_unificado['ID'].duplicated(keep=False).sum()
print(f"Número de duplicados en df_años_unificado['ID']: {num_duplicados_df_años_unificado}")
```

No hay duplicados en ninguno de los dos ID's asi que procedemos a la union de dos df

```python
df_unidos =pd.merge(df_c,df_años_unificado, left_on='id_', right_on='ID', how='left')
```

```python
df_unidos.sample(5)
```

guardamos el nuevo data frame unificado a un nuevo csv

```python
df_unidos.to_csv('../Datos/df_data_unificada.csv', index=False)
```
